```python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(nrows=2, ncols=2, figsize=(12,10))

import matplotlib.pyplot as plt


# Primeiro Gráfico - Linhas ---------------------------------------------------------
meses = ['Jan','Fev','Mar','Abr','Mai','Jun','Jul','Ago','Set','Out','Nov','Dez']
vendas=[85,90,78,92,88,96,94,98,100,102,105,108]

axs[0,0].plot(meses,vendas)
axs[0,0].set_xlabel("Mês")
axs[0,0].set_ylabel("Vendas (10^6 de reais)")
axs[0,0].set_title("Crescimentos das vendas ao longo do ano")

# Segundo Gráfico - Barras ---------------------------------------------------------
produtos = ["Produto A", "Produto B", "Produto C", "Produto D", "Produto E"]
quantidades2 = [23,45,17,30,37]

axs[0,1].bar(produtos,quantidades2)
axs[0,1].set_xlabel("Produto")
axs[0,1].set_ylabel("Quantidade de Vendas")
axs[0,1].set_title("Quantidade de Vendas por Produto")

# Terceiro Gráfico - Pizza ---------------------------------------------------------
categorias = ['Eletrônicos','Roupas','Alimentos','Livros']
quantidades3 = [45,30,15,10]
destaques = [0.1,0,0,0]

axs[1,0].pie(quantidades3, labels=categorias,explode=destaques,autopct='%.1f')
axs[1,0].set_title("Categoria de Produtos Vendidos")

# Quarto Gráfico - Barras Empilhadas ---------------------------------------------------------
anos = [2018,2019,2020,2021]
vendas_online = [5,10,15,20]
vendas_loja = [10,8,6,4]

axs[1,1].bar(anos, vendas_online, label="Online")
axs[1,1].bar(anos,vendas_loja,bottom=vendas_online,label='Loja')
axs[1,1].set_xlabel("Ano")
axs[1,1].set_ylabel("Vendas (10^6 reais)")
axs[1,1].set_title("Vendas Online vs Loja")
axs[1,1].set_xticks(anos)
axs[1,1].legend()

#Ajustes Finais
fig.suptitle("Evolução de Vendas", fontsize=16)
plt.show()
```
