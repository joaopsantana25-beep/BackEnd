Questão 1: Configuração do Ambiente e Criação do Banco

Escreva um script em Python utilizando a biblioteca mysql-connector-
python (ou pymysql) que se conecte ao servidor MySQL, crie um banco de
dados chamado campeonato_futebol e crie as três tabelas (times, jogadores,
partidas) seguindo a modelagem apresentada. Certifique-se de definir
corretamente as chaves primárias (PK) e estrangeiras (FK).

```python
#Conexão Inicial e Banco de Dados
import mysql.connector

conexao = mysql.connector.connect(
    host = "localhost",
    user = "root",
    password = "root"
)

print("Conexão Estabelecida")
```

```python
#Criar Banco

cursor=conexao.cursor()

cursor.execute("CREATE DATABASE IF NOT EXISTS campeonato_futebol")
print("Banco Criado com Sucesso!")
```

```python
#Criando a tabela times
cursor.execute("USE campeonato_futebol")

comando_tabela_times = '''
CREATE TABLE IF NOT EXISTS times(
    id_time INT AUTO_INCREMENT PRIMARY KEY,
    NOME VARCHAR(100) NOT NULL,
    CIDADE VARCHAR(100) NOT NULL)
'''

cursor.execute(comando_tabela_times)
print("Tabela times Criada com Sucesso")
```

```python
#Criando a tabela jogadores
cursor.execute("USE campeonato_futebol")

comando_tabela_jogadores = '''
CREATE TABLE IF NOT EXISTS jogadores(
    id_jogador INT AUTO_INCREMENT PRIMARY KEY,
    id_time INT,
    NOME VARCHAR(100) NOT NULL,
    POSICAO VARCHAR(100) NOT NULL,

    FOREIGN KEY (id_time) REFERENCES times(id_time)
    )
'''

cursor.execute(comando_tabela_jogadores)
print("Tabela jogadores Criada com Sucesso")
```

```python
#Criando a tabela partidas
cursor.execute("USE campeonato_futebol")

comando_tabela_partidas = '''
CREATE TABLE IF NOT EXISTS partidas(
    id_partida INT AUTO_INCREMENT PRIMARY KEY,
    id_mandante INT NOT NULL,
    id_visitante INT NOT NULL,
    gols_mandante INT NOT NULL,
    gols_visitante INT NOT NULL,
    data_partida DATE NOT NULL,

    FOREIGN KEY (id_mandante) REFERENCES times(id_time),
    FOREIGN KEY (id_visitante) REFERENCES times(id_time)
    )
'''

cursor.execute(comando_tabela_partidas)
print("Tabela partidas Criada com Sucesso")
```

Questão 3: Registro de Resultados (Update e Insert)

Crie uma função chamada registrar_partida(id_mandante, id_visitante,
gols_mandante, gols_visitante, data). Além de salvar a partida na tabela
partidas, o script deve exibir no console qual time foi o vencedor ou se
houve empate.

```python
#Função registrar partida
def registrar_partida(id_mandante, id_visitante,gols_mandante,gols_visitante,data):
    comando_sql = "INSERT INTO partidas(id_mandante,id_visitante,gols_mandante,gols_visitante,data_partida) values (%s,%s,%s,%s,%s)"

    values = (id_mandante, id_visitante,gols_mandante,gols_visitante,data)

    cursor.execute(comando_sql,values)
    conexao.commit()
    
    if(gols_mandante>gols_visitante):
        print("O time vencedor foi o mandante!")
        return
    elif (gols_visitante>gols_mandante):
        print("O time vencedor foi o visitante!")
        return
    else:
        print("O jogo empatou!")
        return

print("Função Criada com Sucesso")




```

Questão 4: Consulta Avançada com Junção (Select com JOIN)

Escreva um script Python que execute uma consulta SQL para listar o
nome de todos os jogadores, a posição deles e o nome do time ao qual
pertencem. Exiba o resultado formatado na tela.

```python
#Consulta usando join
comando_join = '''
SELECT
    jogadores.nome,
    jogadores.posicao,
    times.nome
FROM jogadores
INNER JOIN times ON jogadores.id_time = times.id_time
ORDER BY times.nome;
'''

cursor.execute(comando_join)
mostrar_comando = cursor.fetchall()

for dado in mostrar_comando:
    print(f"Nome: {dado[0]} | Posição: {dado[1]} | Time: {dado[2]}")
```

Objetivo: Escreva uma consulta SQL que retorne o nome de cada Time e
a média de gols que esse time marcou jogando dentro de casa (como
mandante).

Requisitos da Consulta:
1. Use INNER JOIN para cruzar os dados da tabela de times com a
tabela de partidas.
2. Use a função de agregação AVG para calcular a média de gols
marcados pelo time mandante.
3. Use GROUP BY para agrupar os resultados pelo nome do time.
4. Use ORDER BY para ordenar o resultado final da maior média de
gols para a menor (ordem decrescente).
5. Defina um apelido (alias) para a coluna da média de gols como
media_gols_casa.

```python
#Consulta Avançada

comando_join_avançado = '''
SELECT 
    times.nome,
    AVG(partidas.gols_mandante)
from partidas
INNER JOIN times ON partidas.id_mandante = times.id_time
GROUP BY times.nome
ORDER BY AVG(partidas.gols_mandante) DESC;
'''

cursor.execute(comando_join_avançado)
mostrar_comando = cursor.fetchall()

for i in mostrar_comando:
    print(f'Time: {i[0]} | Média Gols em Casa: {i[1]:.2f}')

```
