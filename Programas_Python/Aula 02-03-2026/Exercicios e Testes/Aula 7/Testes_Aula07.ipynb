```python
import json
import requests

cotacoes = requests.get("https://economia.awesomeapi.com.br/json/all")
cotacoes_dic = cotacoes.json()


print('Euro:{}'.format(cotacoes_dic["EUR"]["bid"]))
print('Dólar:{}'.format(cotacoes_dic["USD"]["bid"]))
print('Bitcoin:{}'.format(cotacoes_dic["BTC"]["bid"]))


def conversao_real(valor_real):
    valor_dolar = float(cotacoes_dic["USD"]["bid"])
    valor_real_dolar = valor_real/valor_dolar
    print(f'{valor_real} reais valem {valor_real_dolar:.2f} dólares')

def conversao_dolares_real(valor_dolares):
    valor_dolar_em_real = float(cotacoes_dic["USD"]["bid"])
    valor_dolares_convertido = valor_dolares*valor_dolar_em_real
    print(f'{valor_dolares} dólares valem {valor_dolares_convertido:.2f} reais')


conversao_real(5.20)


```

```python
import json
import requests
import matplotlib.pyplot as plt

cotacoes_dolar30d = requests.get("https://economia.awesomeapi.com.br/json/daily/USD-BRL/30")
cotacoes_dolar_dic = cotacoes_dolar30d.json()


dias = list(range(1,31))
lista_cotacoes_dolar = [float(item['bid']) for item in cotacoes_dolar_dic]

plt.plot(dias,lista_cotacoes_dolar)
plt.xlabel("Dias")
plt.ylabel("Valor Dólares")
plt.title("Flutuação do Valor do Dólar")
plt.grid(1)
plt.show()


print(lista_cotacoes_dolar)
```
