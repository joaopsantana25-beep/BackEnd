```python
'''
Crie um gráfico de pizza utilizando os dados abaixo: 

labels=["A","B","C","D"]

size=[15,30,45,10]
'''

import matplotlib.pyplot as plt

labels=["A","B","C","D"]

size=[15,30,45,10]

plt.pie(size,labels=labels,autopct='%.1f')
plt.title("Item 1")
plt.show()
```

```python
''' 
Crie um gráfico de barras para os seguintes dados:

categorias = ["Jan","Feb","Mar","Apr"]
values = [10,20,15,25]
'''

import matplotlib.pyplot as plt 

categorias = ["Jan","Feb","Mar","Apr"]
values = [10,20,15,25]

plt.bar(categorias,values)
plt.xlabel("Categoiras")
plt.ylabel("Values")
plt.title("Item 2")
plt.show()
```

```python
'''
Crie um gráfico de linha com os seguintes dados:

x=[1,2,3,4,5]
y=[2,3,5,7,11]
'''

import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y=[2,3,5,7,11]

plt.plot(x,y,marker="o")
plt.grid(True)
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.title("Item 3")
plt.show()
```

```python
'''
Crie um gráfico de linha com dois conjuntos de dados: 

x=[1,2,3,4,5]
y1=[1,4,9,16,25]
y2=[25,16,9,4,1]

'''

import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y1=[1,4,9,16,25]
y2=[25,16,9,4,1]

plt.plot(x,y1,label ="Linha 1",marker="o")
plt.plot(x,y2,label ="Linha 2",marker="o")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.title("Item 10")
plt.grid(True)
plt.legend()
plt.show()
```
