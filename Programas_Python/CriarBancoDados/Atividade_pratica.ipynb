2. Conexão Inicial e Criação do Banco: Estabelecer a comunicação
com o servidor MySQL local (localhost) e criar um banco de dados
chamado glow_cosmeticos.

```python
#Conexão Inicial e Banco de Dados
import mysql.connector

conexao = mysql.connector.connect(
    host="localhost",
    user = "root",
    password = "root"
)

print("Conexão Iniciada")
```

```python
#Criar Banco

cursor = conexao.cursor()

cursor.execute("CREATE DATABASE IF NOT EXISTS glow_cosmeticos")
print("Banco Criado com Sucesso!")
```

3. Modelagem das Tabelas (DDL): Desenvolver a estrutura relacional
contendo duas tabelas interligadas por chave estrangeira:

    o categorias: Com os campos id (chave primária auto-
    incremento) e nome.
    
    o produtos: Com os campos id (chave primária auto-
    incremento), nome, preco, estoque e id_categoria (chave
    estrangeira).

```python
#Criando a tabela de categorias

cursor.execute("USE glow_cosmeticos")

comando_tabela_categorias="""
Create table if not exists categorias(
    id_categoria int auto_increment primary key,
    nome varchar(100) not null
)
"""

cursor.execute(comando_tabela_categorias)
print("Tabela Criada com sucesso!")
```

```python
#Criando a tabela de produtos

comando_tabela_produtos="""
Create table if not exists produtos(
    id_produto int auto_increment primary key,
    nome varchar(100) not null,
    preco decimal(10,2),
    estoque int,
    id_categoria int,

    Foreign key (id_categoria) references categorias(id_categoria)
)
"""

cursor.execute(comando_tabela_produtos)
print("Tabela Criada com sucesso!")
```

8. Criação e Consumo de Views: Criar uma tabela virtual (View) para
simplificar relatórios futuros, unindo os dados de produtos e suas
respectivas categorias, e consumi-la exibindo os resultados.

```python
#Criando View 

cursor.execute(""" 
CREATE OR REPLACE VIEW vw_Produtos_categorias as
    SELECT 
        id_produto,
        produtos.nome,
        produtos.preco,
        produtos.estoque,
        categorias.nome as "categoria"
    from produtos 
    inner join categorias on produtos.id_categoria = categorias.id_categoria;            
""")

print("View Criada com Sucesso")
```

```python
#Usando a view

cursor.execute("SELECT * FROM vw_Produtos_categorias")
view = cursor.fetchall()

print("\n--- Visualização da View ---")

for i in view:
    print(f"Id_produto: {i[0]} | Nome: {i[1]} | Preco: {i[2]} | Estoque: {i[3]} | Categoria: {i[4]}")

```

9. Subquery (Subconsulta): Realizar uma consulta dinâmica que filtre
e retorne apenas os produtos cujo valor de venda seja superior à
média de preço de toda a loja.

```python
cursor.execute("""
    SELECT 
        nome,
        preco,
        estoque
    FROM produtos
    WHERE preco > (SELECT AVG(preco) FROM produtos)
""")

subquery = cursor.fetchall()

print("\n--- Produtos com Preços Maiores que a Média ---\n")

for i in subquery:
    print(f"Nome: {i[0]} | Preço: R$ {i[1]} | Estoque: {i[2]}")
```

10. Stored Procedure: Desenvolver e disparar um procedimento
armazenado diretamente no servidor para automatizar uma rotina de
atualização (ex: ajuste de estoque).

```python
#Criar a procedure AdicionarItens
cursor.execute("""
CREATE PROCEDURE IF NOT EXISTS AdicionarItens(
    IN p_id INT,
    IN quantidade INT
)
BEGIN
    UPDATE produtos
    SET estoque = estoque + quantidade
    WHERE id_produto = p_id;
END
""")

cursor.execute("""
call AdicionarItens(1,20);
""")

print("Procedure Executado!!!")

#Usando a view

cursor.execute("SELECT * FROM vw_Produtos_categorias")
view = cursor.fetchall()

print("\n--- Visualização da View ---")

for i in view:
    print(f"Id_produto: {i[0]} | Nome: {i[1]} | Preco: {i[2]} | Estoque: {i[3]} | Categoria: {i[4]}")
```
