```python
import mysql.connector

conexao = mysql.connector.connect(
    host = 'localhost',
    user = "root",
    password = "root"
)

print("Conexão Criada")



```

```python
cursor = conexao.cursor()

cursor.execute("CREATE DATABASE IF NOT EXISTS borracharia")
print("Banco Criado")

cursor.execute("USE borracharia")
print("Comando 'use empresa' executado com sucesso!")
```

```python
#Criando a tabela de funcionários 
cursor.execute("USE borracharia")
comando_tabela = """
Create table if not exists funcionarios(
    id int auto_increment primary key,
    nome varchar(255) not null,
    cargo varchar(100),
    salario decimal(10,2)
    )
"""

cursor.execute(comando_tabela)
print("Tabela criada com sucesso")
```

```python
sql_insert = "INSERT INTO funcionarios (nome,cargo,salario) values (%s,%s,%s)"
valores = ("Ana Silva", "Desenvolvedora Python", 7500.00)

cursor.execute(sql_insert,valores)
conexao.commit()
print(f"Funcionario(a) {valores[0]} inserido(a) com sucesso!")
```

```python
#1. O comando SQL permanece o mesmo, usando placeholders

comando_sql = "INSERT INTO funcionarios (nome,cargo,salario) values (%s,%s,%s)"

#2. Criamos uma LISTA contendo várias TUPLAS (cada tupla é um novo funcionário)

lista_funcionarios = [
    ("Enzo Oliveira","Desenvolvedor BackEnd",7000.00),
    ("Valentina Costa","UX Designer",6200.00),
    ("Ricardo Santos","Gerente de Projetos",9500.00),
    ("Fernando Lima", "Anaslitas de RH", 4800.00)
]

#3. Executamos todos de uma vez
cursor.executemany(comando_sql, lista_funcionarios)

#4. Confirmamos a transação no banco
conexao.commit()

print(f"{cursor.rowcount} novos registros foram inseridos com sucesso!!!")

      
```

```python
#Update

slq_update = "UPDATE funcionarios SET cargo = %s, salario = %s where nome = %s"
novos_dados = "Desenvolvedoa Sênior",9500.00,"Ana Silva"

cursor.execute(slq_update , novos_dados)
conexao.commit()
print("\nPromoção Registrada com sucesso!")
```

```python
#Delete

slq_delete = "DELETE FROM funcionarios where nome = %s"
alvo = ("Ana Silva",)

cursor.execute(slq_delete, alvo)
conexao.commit()
print("\nRegistro removido com sucesso!")
```

```python
#Fechar Conexao

cursor.close()
conexao.close()
```
