1. Crie uma classe chamada Pessoa com:
 atributo nome
 atributo idade
Depois, crie um método chamado apresentar() que mostre a frase:
Olá, meu nome é Ana e tenho 25 anos.

```python
class Pessoa():

    def __init__(self,nome,idade):
        self.nome=nome
        self.idade=idade

    def apresentacao(self):
        print(f"Olá, meu nome é {self.nome} e tenho {self.idade} anos.")

    
Ana = Pessoa("Ana",25)
Ana.apresentacao()
```

2. Cronstrua uma classe chamada Produto com:
 atributo nome
 atributo preco
Crie um método chamado mostrar_dados() que exiba o nome e o
preço do produto.
Crie um produto chamado &quot;Caderno&quot; com preço 20.0.

```python
class Produto():

    def __init__(self,nome,preco):
        self.nome = nome
        self.preco = preco

    def mostrar_dados(self):
        print(f"Produto: {self.nome} || Preço: R$ {self.preco:.2F}")

Caderno = Produto("Caderno", 20)

Caderno.mostrar_dados()
```

3. Crie uma classe chamada ContaBancaria com:
 atributo titular
 atributo saldo

Crie:
 um método depositar(valor) para somar ao saldo

 um método mostrar_saldo() para exibir o saldo atual

Depois:

 crie uma conta para &quot;Carlos&quot;

 deposite 100

 mostre o saldo

```python
class ContaBancaria():

    def __init__(self,titular,saldo):
        self.titular = titular
        self.saldo = saldo

    def depositar(self, valor):
        if valor<0:
            print("Valor inválido! Operação cancelada")
        else:
            self.saldo+=valor
            print("Operação realizada com sucesso!!!")

    def mostrar_saldo(self):
        print(f"O seu saldo é: R$ {self.saldo:.2f}")


conta_Carlos = ContaBancaria("Carlos",40)

conta_Carlos.depositar(100)
conta_Carlos.mostrar_saldo()
```
